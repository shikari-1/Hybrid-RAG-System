```python
import json
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer

# 1. Load the chunked data
with open('ww2_chunks.json', 'r') as f:
    chunks_data = json.load(f)

# Extract only the text for embedding
texts = [chunk['content'] for chunk in chunks_data]

# 2. Initialize the model
# 'all-MiniLM-L6-v2' is fast and efficient (384 dimensions)
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. Generate Embeddings
print("Encoding chunks... this may take a minute.")
embeddings = model.encode(texts, show_progress_bar=True)

# 4. Normalize for Cosine Similarity
# Cosine similarity is the inner product of normalized vectors
faiss.normalize_L2(embeddings)

# 5. Build the FAISS Index
dimension = embeddings.shape[1]
index = faiss.IndexFlatIP(dimension) # Inner Product index
index.add(embeddings)

# Save the index for future use
faiss.write_index(index, "ww2_index.faiss")
print(f"Indexed {index.ntotal} chunks successfully.")
```
